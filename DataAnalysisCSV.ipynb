# Creating a derived column

```python
orders["revenue"] = orders["price"] * orders["quantity"]
print(f'Added new column "revenue"')

orders[["order_id", "price", "quantity", "revenue"]].head(5)
```

# multiple Conditions
### Saving conditioned data in a variable

```python
filter = (orders["category"] == "Electronics") & \
          (orders["city"] == "Mumbai") & \
            (orders["revenue"] > 2000)
orders[filter][["order_id", "category", "city", "brand", "revenue"]].head()
```

# .isin()
### filtering orders from particular areas

```python
metro = orders[orders["city"].isin(["Mumbai", "Delhi", "Bengaluru"])]
```

### printing number of orders ordered in those cities

```python
print(f'{len(metro)} items ordered in the three biggest metros \n("Mumbai", "Delhi", "Bengaluru")')
```

# showing data relative to categories
### groupby(col).agg( measure,operation.,eg - "sum") or [measure].sum() 

```python
by_category = orders.groupby("category").agg(total_revenue=("revenue", "sum")).sort_values("total_revenue", ascending=False).reset_index()

by_category
```

### You can aggregate several measures at once with .agg()

```python
summary = orders.groupby("category").agg(total_revenue = ("revenue", "sum"),
                                         total_items = ("order_id", "count"),
                                         distinct_orders = ("order_id", "nunique"),
                                         avg_price = ("price", "mean")
                                        )
summary.sort_values("total_revenue", ascending=False)
```

### Top cities where we got highest revenue

```python
top_city = orders.groupby("city").agg(total_revenue = ("revenue","sum"))

top_city.sort_values("total_revenue",ascending = False).head(3)
```

# merging data from other CSVs 

```python
ordr_cust = orders.merge(customers, on="customer_id", how="left", suffixes=("", "_cust"))
allmerged = ordr_cust.merge(products, on="product_id", how="left", suffixes=("", "_prod"))

allmerged[["order_id", "customer_id", "product_id", "segment", "category", "rating", "revenue"]].head(5)
```

```python
 allmerged.groupby("segment")["revenue"].agg(["sum", "mean", "count"])
```

### checking missing values

```python
allmerged.isna().sum()[lambda s: s > 0]
```

#### Fill categorical gaps with a sensible placeholder

```python
allmerged["brand"] = allmerged["brand"].fillna("Unknown")
```

#### Fill a segment gap with the most common value (mode)

```python
allmerged["segment"] = allmerged["segment"].fillna(allmerged["segment"].mode()[0])
```

```python
allmerged.isna().sum().sum() 
```

#### Fixing datatypes

```python
allmerged["category"] =  allmerged["category"].astype("category")
allmerged["segment"] =  allmerged["segment"].astype("category")
allmerged["order_date"] = pd.to_datetime( allmerged["order_date"])

allmerged.info()
```
