# Question 1: Structured Data, CSV Acquisition, and Missing Values

### 1. Theory Context

* **Structured Data Definition:** The Titanic dataset is classified as structured data because it is organized into a standardized tabular format consisting of rows (individual passenger records) and columns (defined attributes like `Age`, `Sex`, `Pclass`, and `Fare`). Each column possesses a specific data type.
* **Supporting Decision-Making:** Analyzing structured data allows for quantitative evaluation and predictive modeling. For example, calculating survival rates based on passenger class (`Pclass`), age, or gender enables decision-makers to identify key risk factors, evaluate safety protocols, and build predictive machine learning models for emergency survival scenarios.

---

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df.drop_duplicates(inplace=True)

missing_values = df.isnull().sum()
print("Missing values per column:\n", missing_values[missing_values > 0])

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Cabin'] = df['Cabin'].fillna("Unknown")
df.dropna(subset=['Embarked'], inplace=True)

print("Remaining missing values:\n", df.isnull().sum())
```

# Question 2: Semi-structured Data, JSON Parsing, and the Data Lifecycle

### 1. Theory Context

* **Position in the Lifecycle:** Data Acquisition and Pre-processing occur immediately after defining the problem statement. They form the foundational data preparation phase required before performing Exploratory Data Analysis (EDA) or model building.
* **Role in Analysis:** Raw JSON data is semi-structured. Converting it into a clean, tabular DataFrame ensures subsequent analysis is built on reliable inputs. Resolving missing values early prevents runtime errors during statistical aggregation and eliminates skew or bias during visualization and modeling.

---

```python
url = "https://raw.githubusercontent.com/vega/vega/master/docs/data/cars.json"
df = pd.read_json(url)

df.info()

df['Miles_per_Gallon'] = df['Miles_per_Gallon'].fillna(0)
df['Horsepower'] = df['Horsepower'].fillna(0)

print(df.isnull().sum())
print("Total missing values:", df.isnull().sum().sum())
```
