### Question 1: Structured Data, CSV Acquisition, and Missing Values Dataset: Titanic Passenger Demographics
(CSV)

Theory Context:
1. Why this is "Structured Data":
   The Titanic dataset consists of data organized into a clear tabular format 
   with defined rows (individual records/passengers) and columns (attributes/features 
   such as Age, Sex, Pclass, Fare). Each feature has a distinct data type 
   (integer, float, categorical string).

2. Decision-Making Support:
   Analyzing structured data allows organizations to perform predictive modeling 
   and quantitative decision-making. For example, identifying correlations 
   between passenger class (Pclass) or sex and survival probability helps build 
   machine learning models (like Logistic Regression or Decision Trees) to predict 
   outcomes or assess risk factors in similar emergency survival scenarios.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df.drop_duplicates(inplace=True)

missing_values = df.isnull().sum()
print("Missing values per column:\n", missing_values[missing_values > 0])

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Cabin'] = df['Cabin'].fillna("Unknown")
df.dropna(subset=['Embarked'], inplace=True)

print("Remaining missing values:\n", df.isnull().sum())
```

### Question 2: Semi-structured Data, JSON Parsing, and the Data Lifecycle Dataset: Vega Cars Dataset (JSON)

1. Theory Context: Data Analysis Lifecycle
In the overall Data Analysis Lifecycle, Data Acquisition and Pre-processing sit right at the foundation after Problem Definition / Goal Setting:

Problem Definition: Understanding the business goal or research question.

Data Acquisition (Collection): Ingestion of data from diverse sources (APIs, databases, web URLs, CSV/JSON files).

Data Pre-processing (Cleaning & Preparation): Involves handling missing values, removing duplicates, fixing data types, and scaling/encoding.

Exploratory Data Analysis (EDA): Discovering trends, patterns, and anomalies.

Data Modeling / Analysis: Applying statistical methods or machine learning models.

Interpretation & Reporting: Communicating insights to stakeholders for decision-making.

```python
url = "https://raw.githubusercontent.com/vega/vega/master/docs/data/cars.json"
df = pd.read_json(url)

df.info()

df['Miles_per_Gallon'] = df['Miles_per_Gallon'].fillna(0)
df['Horsepower'] = df['Horsepower'].fillna(0)

print(df.isnull().sum())
print("Total missing values:", df.isnull().sum().sum())
```

1. Theory Context: Library Overviews

NumPy: Provides vectorization and low-level numerical computing support. It underlies Pandas data structures and enables vectorized math operations, array transformations, and numerical imputation.

Matplotlib & Seaborn: Used in the Exploratory Data Analysis (EDA) stage to visualize distribution patterns, trends, correlations, and outliers via histograms, scatter plots, and heatmaps.

```python
import pandas as pd
import numpy as np

url = "https://github.com/chris1610/pbpython/raw/master/data/sales_data_types.xlsx"
df = pd.read_excel(url)

first_5 = df.iloc[:5]
df_messy = pd.concat([df, first_5], ignore_index=True)

df_cleaned = df_messy.drop_duplicates(keep='first').copy()

df_cleaned['Customer Number'] = df_cleaned['Customer Number'].astype(int)

print("Shape:", df_cleaned.shape)
print("Customer Number dtype:", df_cleaned['Customer Number'].dtype)
```
